#**Statistics**

```python
df['gold_tonnes'] = df['netWgt'] / 1000
gold_summary = df['gold_tonnes'].describe()
gold_summary
```

#**Key Insights**

**1.** The **Mean (64.65)** is approximately 87 times higher than the **Median (0.74)**. This indicates a heavily skewed distribution where most gold transactions are very small (less than 1 tonne), but a few extremely large entries are pulling the average significantly upward.

**2.** The **Maximum value (2,687.09)** is nearly 181 times larger than the **75th percentile (14.85)**. This suggests that the top 25% of the data contains transactions that account for the vast majority of the total gold volume in the dataset.

```python
country_stats = (
    df.groupby('reporterDesc')
      .agg(
          avg_gold=('gold_tonnes', 'mean'),
          total_gold=('gold_tonnes', 'sum'),
          observations=('gold_tonnes', 'count')
      )
      .sort_values('total_gold', ascending=False)
)

country_stats.head(10)
```

#**Key Insights**

**1.** **Switzerland** leads the dataset with a total accumulation of **10,723 tonnes**. Statistically, its average per observation **(1,191 tonnes**) is the highest among the top reporters, cementing its role as a global gold hub.

**2.** The **United Arab Emirates (UAE)** follows closely with a total of **9,134 tonnes**. With an average of **1,014 tonnes** per entry, it shows a highly concentrated and high-value pattern of gold movement compared to other nations.

```python
year_stats = (
    df.groupby('refYear')
      .agg(
          yearly_total=('gold_tonnes', 'sum'),
          yearly_average=('gold_tonnes', 'mean'),
          yearly_std=('gold_tonnes', 'std')
      )
      .reset_index()
)

year_stats
```

#**Key Insights**

**1.** Global gold accumulation in this dataset grew from **12,076 **tonnes **bold text** in 2019 to **17,901** **tonnes** in 2023, a **48.2% increase** over 5 years.

**2.** There was a massive jump between 2021 and 2023. Specifically, the average gold per entry increased from **53.8 tonnes (2021) to 98.9 tonnes (2023)**, showing that gold movements became much larger in volume recently.

```python
brics = ['Brazil', 'Russia', 'India', 'China', 'South Africa']
eu = ['Germany', 'France', 'Italy', 'Spain', 'Netherlands']

df['Bloc'] = df['reporterDesc'].apply(
    lambda x: 'BRICS' if x in brics
    else 'US' if x == 'United States of America'
    else 'EU' if x in eu
    else 'Others'
)
bloc_stats = (
    df.groupby('Bloc')
      .agg(
          avg_gold=('gold_tonnes', 'mean'),
          total_gold=('gold_tonnes', 'sum'),
          volatility=('gold_tonnes', 'std')
      )
      .reset_index()
)

bloc_stats
```

#**Key Insights**

**1.** The **BRICS nations** show a total gold volume of **9,916 tonnes**, which is nearly **4.5 times higher** than the EU's **2,228 tonnes** in this specific dataset.

**2.** The average gold per record for **BRICS is 268 tonnes**, whereas the **EU average is only 85 tonnes**. This suggests that BRICS nations are making much larger individual gold acquisitions/movements than European nations.

```python
recent_gold = (df[df['refYear'].isin([2022,2023])].groupby(['reporterDesc', 'refYear'])['gold_tonnes'].sum().unstack()
)

recent_gold.sort_values(2023, ascending=False).head(10)
```

#**Key Insights**

**1.** The UAE saw an unprecedented **increase of 3,332 tonnes** between 2022 and 2023, making it the fastest-growing gold holder/trader in the recent period.

**2.** Armenia emerged as a surprise top gainer with a **482 tonne increase**, outperforming major players like Australia and China in terms of net growth for that specific window.

#**Data Visualization**

```python
df['gold_tonnes'] = df['netWgt'] / 1000
df['Year'] = pd.to_datetime(df['period']).dt.year
```

```python
avg_country_gold = (
    df.groupby('reporterDesc')['gold_tonnes']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

avg_country_gold.plot(kind='bar', figsize=(10,5))
plt.title("Average Gold Purchase per Country")
plt.ylabel("Gold (tonnes)")
plt.xlabel("Country")
plt.show()
```

#**Key Insights**

**1.** The visualization shows a massive lead for **Switzerland ($1,191.49$ tonnes avg)** and the **United Arab Emirates ($1,014.94$ tonnes avg)**. This indicates that gold movement is not distributed evenly across all nations

**2.** There is a significant "step-down" after the top three countries. While the leaders average over **$1,000$ tonnes**, the next tier **(Hong Kong and China) drops to the $500$-$600$ tonne** range. This visual contrast highlights the gap between major global gold markets and secondary national reserves.

```python
yearly_gold = (
    df.groupby('refYear')['gold_tonnes']
      .sum()
      .sort_index()
)

plt.figure(figsize=(10,5))
plt.plot(yearly_gold.index, yearly_gold.values, marker='o')
plt.title("Year-wise Gold Accumulation Trend")
plt.xlabel("Year")
plt.ylabel("Gold (tonnes)")
plt.grid(True)
plt.show()
```

#**Key Insights**

**1.** The line chart shows a sharp "V-shaped" recovery. After a dip in **2020 ($10,733.96$ tonnes)**, the trend sky-rocketed to **$17,901.08$** tonnes by 2023.

**2.** The trend between 2021 and 2023 shows a nearly linear and steep upward trajectory. This suggests that the "Increasing Trend" you were looking for is not just a temporary spike

```python
bloc_gold = (
    df.groupby(['Bloc', 'period'])['gold_tonnes']
      .sum()
      .reset_index()
)

for bloc in ['BRICS', 'US', 'EU']:
    data = bloc_gold[bloc_gold['Bloc'] == bloc]
    plt.plot(data['period'], data['gold_tonnes'], label=bloc)

plt.title("Gold Accumulation: BRICS vs US vs EU")
plt.xlabel("Year")
plt.ylabel("Gold (tonnes)")
plt.legend()
plt.grid(True)
plt.show()
```

#**Key Insights**

**1.** The visualization reveals that BRICS nations maintained a high baseline of gold volume **(peaking at $2,364.27$ tonnes in 2021)**. Visually, their line remains consistently above the EU line, highlighting a strategic shift where emerging economies are prioritizing gold more aggressively than traditional European powers.

**2.** The most dramatic visual in this chart is the line for "Others." In 2023, this category hit **$15,150.96$ tonnes.**

```python
recent = df[df['refYear'].isin([2022, 2023])]

change = (
    recent.groupby(['reporterDesc', 'refYear'])['gold_tonnes']
    .sum()
    .unstack()
)

change['Increase'] = change[2023] - change[2022]

top_increase = change['Increase'].sort_values(ascending=False).head(10)

top_increase.plot(kind='bar', figsize=(10,5))
plt.title("Top Gold Reserve Increases (2022–2023)")
plt.ylabel("Increase in Gold (tonnes)")
plt.xlabel("Country")
plt.show()
```

#**Key Insights**

**1.** This bar chart is dominated by a single massive bar for the **United Arab Emirates**, which saw a net increase of **$3,332.51$ tonnes** in just one year.

**2.** Beyond the major hubs, the chart highlights unexpected growth in regions like **Armenia ($+482.93$ tonnes) and Uganda ($+113.36$ tonnes)**

```python
plt.figure(figsize=(8,5))
plt.hist(df['gold_tonnes'], bins=30)
plt.title("Distribution of Gold Purchases")
plt.xlabel("Gold (tonnes)")
plt.ylabel("Frequency")
plt.show()
```
