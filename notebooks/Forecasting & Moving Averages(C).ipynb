```python
gold_df  = GOLD_FILE
crude_df = CRUDE_FILE
btc_df   = BTC_FILE
```

```python
gold_yearly = (
    gold_df
    .groupby("refYear", as_index=False)["primaryValue"]
    .sum()
    .sort_values("refYear")
)

gold_yearly["MA_3Y"] = gold_yearly["primaryValue"].rolling(3).mean()
gold_yearly.tail()
```

```python
plt.figure()
plt.plot(gold_yearly["refYear"], gold_yearly["primaryValue"], label="Actual")
plt.plot(gold_yearly["refYear"], gold_yearly["MA_3Y"], label="3-Year Moving Avg")
plt.xlabel("Year")
plt.ylabel("Gold Trade Value")
plt.title("Gold – 3-month Moving Average Forecast")
plt.legend()
plt.savefig(OUT_DIR / "gold_forecast.png", dpi=200)
plt.show()
```

#**2.**

The 3-year moving average smooths short-term fluctuations and confirms the long-term growth pattern in gold trade.The increase will be upto 180 Billion USD.

```python
crude_yearly = (
    crude_df
    .groupby("refYear", as_index=False)["primaryValue"]
    .sum()
    .sort_values("refYear")
)

crude_yearly["MA_3Y"] = crude_yearly["primaryValue"].rolling(3).mean()
crude_yearly.tail()
```

```python
plt.figure()
plt.plot(crude_yearly["refYear"], crude_yearly["primaryValue"], label="Actual")
plt.plot(crude_yearly["refYear"], crude_yearly["MA_3Y"], label="3-Year Moving Avg")
plt.xlabel("Year")
plt.ylabel("Crude Oil Trade Value")
plt.title("Crude Oil – 3- month Moving Average Forecast")
plt.legend()
plt.savefig(OUT_DIR / "crude_forecast.png", dpi=200)
plt.show()
```

```python
btc_df["Time"] = pd.to_datetime(btc_df["Time"], errors="coerce")

value_cols = [c for c in btc_df.columns if c != "Time"]
btc_df[value_cols] = btc_df[value_cols].apply(pd.to_numeric, errors="coerce")

btc_df["TotalVolume"] = btc_df[value_cols].sum(axis=1)
btc_df["Year"] = btc_df["Time"].dt.year

btc_yearly = (
    btc_df
    .groupby("Year", as_index=False)["TotalVolume"]
    .sum()
    .sort_values("Year")
)

btc_yearly["MA_3Y"] = btc_yearly["TotalVolume"].rolling(3).mean()
btc_yearly.tail()
```

```python
plt.figure()
plt.plot(btc_yearly["Year"], btc_yearly["TotalVolume"], label="Actual")
plt.plot(btc_yearly["Year"], btc_yearly["MA_3Y"], label="3-Year Moving Avg")
plt.xlabel("Year")
plt.ylabel("Bitcoin Trading Volume")
plt.title("Bitcoin – 3- month Moving Average Forecast")
plt.legend()
plt.savefig(OUT_DIR / "bitcoin_forecast.png", dpi=200)
plt.show()
```

#**Key Insights**

#**1.**
The actual Bitcoin trading volume shows extreme volatility over the observed period. Trading volume rises sharply from nearly zero in 2010.This represents an explosive increase of over 1 billion units within a few years, indicating a period of intense speculative activity.After this peak, trading volume declines steeply by 2018-2019.

#**2**
The 3-year moving average smooths short-term spikes and reveals the underlying trend. The moving average increases.After 2017, the moving average consistently declines.Bitcoin trading activity stabilized at much lower levels in later years.
